```python
#-------> PROJECT 1 Data Visualization
#Import required libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

#Load the dataset
df = sns.load_dataset("iris")
#Display the first few rows
print(df.head())
# Basic info
print(df.info())
#Summary statistics
print(df.describe())
#pair plot relationship
sns.pairplot(df, hue="species",corner=True,palette="Set2")
plt.suptitle("Pairplot of Iris Features by Species",y=1.02)
plt.show()

#Box plot
plt.figure(figsize=(12,6))
sns.boxplot(data=df,x="species",y="sepal_length",palette="pastel")
plt.title("Sepal Length Distribution by Species")
plt.show()

#Histogram plot
plt.figure(figsize=(10,5))
sns.histplot(data=df,x="petal_length",kde=True,bins=20,color="skyblue")
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Count")
plt.show()

#viloin plot
plt.figure(figsize=(12,6))
sns.violinplot(data=df,x="species",y="petal_length",palette="muted")
plt.title("Petal Length Density by Species")
plt.show()

# Correlation heatmap 
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Iris Features")
plt.show()
```
